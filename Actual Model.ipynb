```python
# AQI-related features
features = ['PM2_5', 'PM10', 'NO2', 'SO2', 'O3', 'Temperature', 'Humidity', 'WindSpeed']
x = df[features]
y = df['AQI']
# Target for Health Impact
y_health = df['HealthImpactScore']
```

```python
!pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test, y_health_train, y_health_test = train_test_split(x, y, y_health, test_size=0.2, random_state=38)
```

```python
# Train AQI model
from sklearn.ensemble import RandomForestRegressor
rf_aqi = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_aqi.fit(x_train, y_train)
```

```python
# Predict
predicted_aqi = rf_aqi.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np
print("AQI Model Performance:")
print("MSE:", mean_squared_error(y_test, predicted_aqi))
print("RMSE:", np.sqrt(mean_squared_error(y_test, predicted_aqi)))
print("MAE:", mean_absolute_error(y_test, predicted_aqi))
print("R2 Score:", r2_score(y_test, predicted_aqi))
```

```python
import joblib

joblib.dump(rf_aqi, "aqi.pkl")
```

```python
import pandas as pd

x_health = pd.DataFrame({
    'PM2_5': x_test['PM2_5'],
    'PM10': x_test['PM10'],
    'NO2': x_test['NO2'],
    'SO2': x_test['SO2'],
    'O3': x_test['O3'],
    'Temperature': x_test['Temperature'],
    'Humidity': x_test['Humidity'],
    'Wind_Speed': x_test['WindSpeed'],
        'Predicted_AQI': predicted_aqi
})
y_health = y_health_test
```

```python
## Scale Health Impact features
from sklearn.preprocessing import StandardScaler
scaler_health = StandardScaler()
x_health_scaled = scaler_health.fit_transform(x_health)
```

```python
## Train Health Impact Model
rf_health = RandomForestRegressor(n_estimators=100, random_state=38)
rf_health.fit(x_health_scaled, y_health)
```

```python
# Predict
yh_pred = rf_health.predict(x_health_scaled)
```

```python
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np
print("Health Impact Model Performance:")
print("MSE:", mean_squared_error(y_health, yh_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_health, yh_pred)))
print("MAE:", mean_absolute_error(y_health, yh_pred))
print("R2 Score:", r2_score(y_health, yh_pred))
```

```python
joblib.dump(rf_health, "health.pkl")
joblib.dump(scaler_health, "health_scaler_model.pkl")
```
